#1) Quyida berilgan datasetni yuklab olib, ustun nomlarini koʻring va ularni maʼno jihatdan tushunarli shaklga keltiring. `Unnamed: 0` ustunini tashlab yuboring!

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url)
df.drop('Unnamed: 0', axis=1, inplace=True)
df.head()
```

```python
#javobni shu yerda yozing
df.rename(lambda x: x.lower(),axis=1, inplace=True)
df.rename(columns={
    'order id':'order_id',
    'customer id':'customer_id',
    'order total':'order_total',
    'order date':'order_date',
    'payment method':'payment_method'}, inplace=True)
df.head()
```

#2) Maʼlumotlarning umumiy holatini (info(), describe()) tekshiring.

```python
#javobni shu yerda yozing
df.info()
```

#3) Har bir ustundagi boʻsh qiymatlarni aniqlang va ularni toʻldirish strategiyasini ishlab chiqing.

```python
#javobni shu yerda yozing
# 1 Dastlab category ustunidagi NaN qiymatlarni aniqlab ularni tashlab yuboramiz
df.dropna(subset=['category'], inplace=True)

# 2 Endi item ustunidagi NaN qiymatlarini o'zidan oldingi qiymatga o'zgartiramiz
df['item'] = df['item'].fillna(method='ffill')

# 3 quantity va order total ustunlarini ham o'zgartiramiz
# bu ikki ustun bir xel NaN qiymatlari mavjud emas!
# Shuning uchun har ikki ustunga bir xil ozidan oldingi qiymatlarni berib chiqamiz
df['quantity'] = df['quantity'].fillna(method='ffill')
df['order_total'] = df['order_total'].fillna(method='ffill')

# 4 Oxirgi payment_method ustunidagi NaN qiymatlarini Cash o'zgartiramiz
df['payment_method'] = df['payment_method'].fillna('Cash')
df.sample(10)
```

```python
df.isnull().sum() # hamma ustunlarni tekshirib NaN qiymati qolmaganini tekshirib olamiz!
```

#4) Har bir ustundagi takroriy qiymatlarni toping va ularga qanday oʻzgartirish kiritish kerakligini hal qiling.

```python
#javobni shu yerda yozing
df['order_id'].duplicated().sum() # bu normal holat
df['customer_id'].duplicated().sum() # bu normal holat sababi mijozlarni ber necha bor tashrif buyurishi mumkin
```

#5) Maʼlumotlar toʻplamidagi noto‘g‘ri maʼlumotlar (masalan, son turishi kerak bo‘lgan ustunda matn bo‘lsa) mavjudligini tekshiring va to‘g‘rilang.

```python
# 'price' ustunini tekshiramiz
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')
df['order_total'] = pd.to_numeric(df['order_total'], errors='coerce')
df.info()
```

#6) Maʼlumotlar toʻplamidagi **katta/kichik harf** muammolari boʻlgan joylarni toping va bir xil formatga keltiring.

```python
#javobni shu yerda yozing
df['item'] = df['item'].map(lambda x: x.capitalize())
df['item'].unique()
```

#7) Maʼlumotlar turini oʻzgartirish.
 * Ustunlarning maʼlumot turlarini tekshiring. Notoʻgʻri turlarga ega boʻlgan ustunlarni toʻgʻri formatga oʻtkazing.
 * `order_date` yoki unga oʻxshash ustunni datetime formatiga oʻtkazing.

```python
#javobni shu yerda yozing
df['order_date'] = pd.to_datetime(df['order_date'], format="%Y-%m-%d")
df[['order_date']].info()
```

#8) Mahsulot nomlarida noto‘g‘ri yoki nomaʼlum belgilar (?, NaN, #) mavjudligini tekshiring va ularni tuzating.

```python
#javobni shu yerda yozing
mask = df['item'].str.contains(r'[?#]|NaN', na=False, regex=True)
df = df[~mask]
df['item'].unique()
```

#**Maʼlumotlarni filtrlash va qayta ishlash**

#9) Manfiy narx yoki 0 ga teng narx qiymatlarini toping va ularni qanday oʻzgartirish kerakligini hal qiling.

```python
#javobni shu yerda yozing
# Tekshirish
mean_price = df['price'].mean()
df.loc[df['price'] <= 0, 'price'] = mean_price
df['price']
```

#10) Boʻsh toifalar (kategoriya) mavjudligini tekshiring va toʻgʻri kategoriya berish strategiyasini ishlab chiqing.

```python
#javobni shu yerda yozing
df.isnull().sum() # bo'sh kategoriya mavjud emas!
```

#11) Bir xil, lekin turlicha yozilgan qiymatlar (Coffee, cofee, COFEE) mavjudligini tekshiring va tuzating.

```python
#javobni shu yerda yozing
df["item"] = (
    df["item"]
    .str.strip()                          # bosh/oxiridagi bo'shliqlar
    .str.capitalize()                          # hammasini birinchi harf katta bo'ladi
    .str.strip()                          # yana bo'shliq tozalash
)
```

#12) `Customer ID` yoki boshqa identifikatorlarning takrorlanishini tekshiring.

```python
#javobni shu yerda yozing
# Agar bir xil Order ID ikki marta kelsa — BU xato:
df.duplicated(subset=['order_id']).sum()  # → 0  Hammasi yaxshi

# Agar butun qator aynan bir xil bo'lsa — BU xato:
df.duplicated().sum()  # → 0  Hammasi yaxshi
```

#13) Mahsulotning ortiqcha taʼriflarini olib tashlang (Water (0.5), Water (1.0), Water → faqat Water qoldirish).

```python
#javobni shu yerda yozing
df['item'] = (df['item']
  .str.replace(r"[#?]", "", regex=True) # # va ? belgisini olib tashlash
  .str.replace(r"\(.*?\)", "", regex=True) # (0.5), (1.0) kabi qavslarni olib tashlash
)
df['item'].unique()
```
